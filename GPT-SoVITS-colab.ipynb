```python
import subprocess, time

# whisperをインストール
!pip install openai-whisper pydub -q

import whisper
from pydub import AudioSegment

# ファイルパスを確認
src = "/content/忍(完全体)WAV形式.wav"

# 最初の7秒だけ切り出す
audio = AudioSegment.from_wav(src)
clip = audio[:7000]  # 7秒
clip_path = "/content/kisshot_ref.wav"
clip.export(clip_path, format="wav")
print(f"✅ 切り出し完了: {clip_path}")

# Whisperで文字起こし
print("文字起こし中...（1〜2分かかる）")
model = whisper.load_model("medium")
result = model.transcribe(clip_path, language="ja")
text = result["text"]
print("=" * 50)
print("✅ 文字起こし結果:")
print(text)
print("=" * 50)
print("\n↑ この文字列をWebUIの『Text for reference audio』にコピペ")
```
